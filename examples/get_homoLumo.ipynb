# HOMO/LUMO extraction

```python
from aiida.orm import load_node
from aiida import load_profile
from pathlib import Path
from ase.io import read
from ase.visualize import view
import os
import pandas as pd
load_profile()
```

```python
# Get order of salts
salt_order = [
    '236-methylphenol.xyz', '24-methylphenol.xyz', 
    '246-methylphenol.xyz', '26-methylphenol.xyz', 
    '3-methoxylphenol.xyz', '34-methylphenol.xyz', 
    '35-methylphenol.xyz', 'phenol.xyz'
]
```

```python
pks = [5385, 5388, 5391, 5394, 5397, 5400, 5403, 5406]
```

```python
def get_homo(pk:int):
    wc = load_node(pk)
    results = wc.outputs.output_parameters.get_dict()
    eigs = results["moenergies"]
    return eigs[-1][results["homos"][0]]

def get_lumo(pk:int):
    wc = load_node(pk)
    results = wc.outputs.output_parameters.get_dict()
    eigs = results["moenergies"]
    return eigs[-1][results["homos"][0]+1]

def view_atoms(pk:int):
    wc = load_node(pk)
    atoms = wc.outputs.relaxed_structure.get_ase()
    view(atoms)
```

```python
homos = []
lumos = []
names = []
for ipk, pk in enumerate(pks):
    homo = get_homo(pk)
    lumo = get_lumo(pk)
    name = salt_order[ipk]
    homos.append(homo)
    lumos.append(lumo)
    names.append(name)
    if ipk == 4:
        view_atoms(pk)

df = pd.DataFrame({"name": names, "homo": homos, "lumo": lumos})
print(df.T)
```
