```python
from aiida.orm import load_node
from aiida import load_profile
from ase.units import Ha
import numpy as np
load_profile()
```

```python
def get_assoc_energy(pk:int):
    n = load_node(pk)
    an = n.inputs.an_params.get_dict()["freeenergy"]
    cat = n.inputs.cat_params.get_dict()["freeenergy"]
    pair = n.inputs.pair_params.get_dict()["freeenergy"]
    return (pair - (an + cat))

def get_assoc_dft_energy(pk:int):
    n = load_node(pk)
    an = n.inputs.an_params.get_dict()["scfenergies"][-1]
    cat = n.inputs.cat_params.get_dict()["scfenergies"][-1]
    pair = n.inputs.pair_params.get_dict()["scfenergies"][-1]
    return (pair - (an + cat))
```

```python
pf6 = get_assoc_energy(5082) * Ha
# boph = get_assoc_energy(5102) * Ha
# bph = get_assoc_energy(5107) * Ha
boph = get_assoc_energy(5211) * Ha
bph = get_assoc_energy(5216) * Ha
bhfip = get_assoc_energy(5115) * Ha

pf6_scf = get_assoc_dft_energy(5082)
boph_scf = get_assoc_dft_energy(5102)
bph_scf = get_assoc_dft_energy(5107)
bhfip_scf = get_assoc_dft_energy(5115)

print(f"pf6: {pf6}, boph: {boph}, bph: {bph}, bhfip: {bhfip}")
print(f"pf6: {pf6_scf}, boph: {boph_scf}, bph: {bph_scf}, bhfip: {bhfip_scf}")
print(f"pf6: {pf6_scf - pf6}, boph: {boph_scf - boph}, bph: {bph_scf - bph}, bhfip: {bhfip_scf - bhfip}")
```

```python
def na_b_dist(struct_pk):
    atoms = load_node(struct_pk).get_ase()
    na_idx = atoms.get_chemical_symbols().index("Na")
    b_idx = atoms.get_chemical_symbols().index("B")
    return np.linalg.norm(atoms.positions[na_idx,:] - atoms.positions[b_idx, :])

```

```python
boph = na_b_dist("2a54db46")
bph = na_b_dist("8c840713")

print(f"Na - B distance for BOPh: {boph}, BPh: {bph}")
```
